## 3. Train Model

We use `yolov8n.pt` (Nano) for the smallest file size and fastest web inference.

```python
# Update 'data.yaml' path if needed. It usually sits in the downloaded dataset folder.
# Example: data='/content/Billboard-Detection-1/data.yaml'

model = YOLO('yolov8n.pt')  # load a pretrained model

# Train the model
# imgsz=640 is standard. epochs=50 is usually enough for transfer learning.
results = model.train(data='/content/dataset/data.yaml', epochs=50, imgsz=640)
```

## 4. Export to ONNX

This creates the `.onnx` file needed for the web app.

```python
# Export the model
success = model.export(format='onnx')
```
